```python
# =========================================================
# Netflix Content Analysis and Clustering using Python
# =========================================================



# 1. Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder
from sklearn.cluster import KMeans


# 2. Load Dataset


df = pd.read_csv("netflix_titles.csv")

print("FIRST 5 ROWS")
print(df.head())


# 3. Basic Information


print("\nDATASET SHAPE")
print(df.shape)

print("\nDATASET INFO")
print(df.info())



# 4. Check Missing Values


print("\nMISSING VALUES")
print(df.isnull().sum())



# 5. Handle Missing Values


df['director'].fillna('Unknown')
df['cast'].fillna('Unknown')
df['country'].fillna('Unknown')
df['rating'].fillna(df['rating'].mode()[0])

# Convert date column safely
df['date_added'] = pd.to_datetime(
    df['date_added'],
    errors='coerce'
)

# Remove rows where date conversion failed
df.dropna(subset=['date_added'], inplace=True)



# 6. Remove Duplicates


df.drop_duplicates(inplace=True)


# 7. Feature Engineering


# Create Year Added Column
df['year_added'] = df['date_added'].dt.year



# 8. Exploratory Data Analysis (EDA)


# Movies vs TV Shows


plt.figure(figsize=(6,4))

sns.countplot(
    x='type',
    data=df
)

plt.title("Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Count")

plt.tight_layout()
plt.show()



# Top 10 Countries Producing Netflix Content


top_countries = df['country'].value_counts().head(10)

plt.figure(figsize=(10,5))

sns.barplot(
    x=top_countries.index,
    y=top_countries.values
)

plt.xticks(rotation=45)

plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Country")
plt.ylabel("Count")

plt.tight_layout()
plt.show()



# Ratings Distribution


plt.figure(figsize=(10,5))

sns.countplot(
    y='rating',
    data=df,
    order=df['rating'].value_counts().index
)

plt.title("Content Ratings Distribution")
plt.xlabel("Count")
plt.ylabel("Rating")

plt.tight_layout()
plt.show()


# Netflix Release Trend


release_trend = (
    df['release_year']
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(12,5))

plt.plot(
    release_trend.index,
    release_trend.values,
    marker='o'
)

plt.title("Netflix Content Release Trend")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.tight_layout()
plt.show()



# 9. Clustering Analysis


# Select required columns
cluster_df = df[['type', 'release_year']].copy()

# Encode categorical values
encoder = LabelEncoder()

cluster_df['type'] = encoder.fit_transform(
    cluster_df['type']
)

# Create KMeans Model
kmeans = KMeans(
    n_clusters=3,
    random_state=42
)

cluster_df['Cluster'] = kmeans.fit_predict(
    cluster_df
)



# 10. Correlation Heatmap


plt.figure(figsize=(5,4))

sns.heatmap(
    cluster_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()



# 11. Cluster Visualization


plt.figure(figsize=(8,5))

sns.scatterplot(
    x=cluster_df['release_year'],
    y=cluster_df['type'],
    hue=cluster_df['Cluster'],
    palette='Set2'
)

plt.title("K-Means Clustering of Netflix Content")
plt.xlabel("Release Year")
plt.ylabel("Encoded Content Type")

plt.tight_layout()
plt.show()






# 12. Save Cleaned Dataset


df.to_csv(
    "cleaned_netflix_data.csv",
    index=False
)

print("\nCleaned dataset saved successfully.")


```
